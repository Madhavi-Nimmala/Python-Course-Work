write() function

```python
#To wite the content inside the file
#We will use the "W" mode with write() function
#If file dont exist then "W" MODE Will create a new file.
#Read mode "r" doesn't create the file
```

```python
with open("madhu.txt", "w") as f:
    f.write("present madhu is under depression she need peace and happiness")
```

```python
with open("madhu.txt","r") as f:
     print(f.read())
```

```python
#in "W" mode data get overwrite on previous data
with open("madhu.txt","w") as f:
       f.write("madhu bestfriends are jaanu hansika")
```

```python
with open("madhu.txt","r") as f:
     print(f.read())
```

```python
#append mode:a
```

```python
with open("madhu.txt","a") as f:
    f.write("hansika is going to dubai after 4-5 months")
```

```python
with open("madhu.txt","r") as f:
     print(f.read())
```

```python
#Writelines: Writing multiple lines together in the file
```

```python
with open("madhu.txt","a") as f:
     f.writelines(["Hansika is going to dubai after 4-5 months.\n", "And she has a bright future"])
```

```python
with open("madhu.txt","r") as f:
     print(f.read())
```

```python
#w+mode:write/read mode
#w+mode will clear the previous content
#w+ create new file if file doesn't exist

with open("madhavi.txt","w+") as f:
    f.write("This is new file")
    print(f.read())
```

```python
#w+mode:write/read mode
#w+mode will clear the previous content
#w+ create new file if file doesn't exist

with open("madhavi.txt","w+") as f:
    f.write("This is new file")
    f.seek(0)#take the head here and their(end to begin) stat read untill stop(end) and again start read from begining
    print(f.read())
```

```python
#w+mode:write/read mode
#w+mode will clear the previous content
#w+ create new file if file doesn't exist

with open("madhavi.txt","w+") as f:
    f.write("This is new file")
    f.seek(4)#take the head here and their(end to begin) start read untill stop(end) and again start read from begining
    print(f.read())
```

```python
#r+mode: read/write mode
with open("hansika.txt","r+") as f:
    f.write("hansika is a naughty girl")
    f.seek(0)
    print(f.read())
```

```python
#r+mode: read/write mode
with open("madhavi.txt","r+") as f:
    f.write("hansika is a naughty girl.")
    f.seek(0)
    print(f.read())
```

```python
#a+mode: Append/read
with open("madhavi.txt","a+") as f:
    f.write("This is new content")
    f.seek(0)
    print(f.read())
```
