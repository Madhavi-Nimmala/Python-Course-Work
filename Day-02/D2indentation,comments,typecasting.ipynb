```python
# Integer typecasting:int()
v1 = int(5.6)#float to int
v2 =int("300")#string to int
v3= int(True)#boolean to int
v4=int(False)#boolean to int
v5=int("101",2)#binary to int- binary values should have(1,2)
print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
print(v4,type(v4))
print(v5,type(v5))
```

```python
#convert int to float 
v1=float(23)
v2=float("101")
v3=float(True)
print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
print(v4,type(v4))
```

```python
#convert int to string
v1=str(23)
v2=str(10.1)
v3=str(True)
print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
```

```python
#convert to boolean:bool()
v1=bool(23)
v2=bool("101")
v3=bool(True)
v4=bool(False)
v5=bool("hello")
print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
print(v4,type(v4))
print(v5,type(v5))
      
```
