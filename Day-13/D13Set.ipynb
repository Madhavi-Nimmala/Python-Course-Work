#Creating different types of set

```python
empty_set=set()
set2=set([12,45,67])
set3=set((10,20,30))
#frozenaet is immutable(cannot perform insert,delete,update)
fro_set=([10,20,30,40,50])
nested_set={(10,20,30),"hello",30}#tuple+frozenset can be store in the set
nested_set1={[10,20,30],"hello",30}#not allowed to store the list in set
nested_set2={{10,20,30},"hello",30}#set inside set cannot be store
#imp:only immutable items can be stored in the set(tuple&frozenset),but mutable items(list,set)cannot be store
print(empty_set)
print(set2)
print(set3)
print(fro_set)
print(nested_set)
print(nested_set1)
print(nested_set2)
```

#INSET IN SET (ADD(),UPDATE()

```python
set1={10,20.5,"hello"}
#add()function is used to insert one item in the set
set1.add("world")
print(set1)
```

```python
set1.add((10,20,30,40))
print(set1)
```

```python
#UPDATE IS USED FOR INSERTING MULTUIPLE ITEMS TOGETHER
set1.update((1000,2000))
print(set1)
```

```python
set1.update({(1000,2000)})
print(set1)
```

```python
#multiple items can be passed to update()function in a set or tuple or list
s1={101,102}
t1=(201,202)
l1=[301,302]
set1.update(s1)
set1.add(t1)
print(set1)
```

```python
#Delete operation on set(remove(),discard(),pop(),clear(),del)
dinner_set={"bagara rice","biryani","papu","chapati","fish","mutton","prawns"}
#discard()function delete the one item whose name is provided 
#if the item is not their already remove again. If you want to remove the same ones it shows the error
dinner_set.remove("prawns")
#discard()function delete the one item whose name is provided 
#if the item is not their it shows the no error
dinner_set.discard("biryani")
print(dinner_set)
```

```python
dinner_set.pop()
print(dinner_set)
```

```python
dinner_set.clear()
print(dinner_set)
```

```python
del dinner_set
```
