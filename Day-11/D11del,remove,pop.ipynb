delete item from list(remove(),pop(),clear(),del()

```python
food_list=["papu","biryani","fried rice","sambar", "curd rice","payasam"]
food_list.remove("sambar")
food_list.pop(1)#pop is used to delete the item based on the index or witoutnindex it will delete the last element
food_list.pop()

print(food_list)
del food_list[1:4]
print(food_list)
food_list.clear()
print(food_list)
del food_list
      
```

```python
#popular list inbi=uilt function
age_list=[22,23,22,20,19,18,21,16,19,20,99]

print("length:",len(age_list))
print("maximum:",max(age_list))
print("minimum:",min(age_list))
print("sum:",sum(age_list))
#in python sort() function use the TIM sort
#Tim sort=>(Merge+Insert)
#Tim sort=>O(nlogn)
age_list.sort()
print("Ascending order:", age_list)

age_list.sort(reverse=True)
print("Descending order:", age_list)
```
