```python
obj = lambda a : a + 100

print(obj(10))
```

```python

obj = lambda a,b : a + b

print(obj(100, 200))
```

```python

obj1 = lambda a,b,c : a*b*c

val = obj1(1,2,3)
print(val)
```

```python
def fun(p):
    return lambda a : a**p

obj1 = fun(10)

print(obj1(2))
print(obj1(3))
print(obj1(4))
```

```python

# using lambda function inside the other function
def fun(p):
    obj = lambda a : a**p if a%2==0 else a+p
    return obj

obj = fun(2)
print(obj1(2))
print(obj1(3))
print(obj1(4))
```

```python
# sort the below given list of tuple w,r,t marks
# lambda function popularly used with sorted() function

students = [("pavan", 80), ("dinesh", 75), ("chander" , 78)]

sorted_data = sorted(students, key = lambda a : a[1] )
print(sorted_data)
```

```python
students = [("pavan", 80, 5), ("dinesh", 75, 100), ("chander" , 78, 15)]

sorted_data = sorted(students, key = lambda a : a[2] )
print(sorted_data)
```

```python
students = {"pavan": 80, "dinesh": 75, "chander" : 78}
data = students.items()   # converting dictionary to list of tuple
# print(data)
sorted_data = sorted(data, key = lambda a : a[1])
print(sorted_data)
```
