```python
#Example-3:Continue statement
#Continue will take control to the start of the loop
i=1
while i<10:
    i+=1
    if i<5:
        continue
    print(i)
```

```python
#Example-4: else statement with while loop
#else statmet executes only if loop is sucessfully completed
i=1
while i<=10:
  print(i)
  i+=1
  if i>5:
    break
else:
  print("go to hell")
```

```python
#Example-5:pass statment
# A pass statement is a placeholder for future code
i=0
while i<=10:
    if i==5:
      pass
    else:
     print(i)
    i+=1
```
