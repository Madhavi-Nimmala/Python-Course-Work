```python
#eample-1 structure of class
#class is a keyword,food is the class name
#class variable value cannot be change by an object
#class variable value can be only  change by class name
class food:
    food_name="biriyani" #class variable
obj3=food()
obj3.food_name="papu"
print(obj3.food_name)#printing local variable to obj3
obj1 = food()#creating the object(obj) of food() class
print(obj1.food_name) #accessing the class variable
food.food_name="mutton"
obj2=food()
print(obj2.food_name) #class object (obj2) can acess the class data
print(food.food_name) #access the class variable with class name
```

```python
#example-2 constructor function
#constructor function automatically get called when an object is created
#constructor function is used for initialising the variables
#instance variable value is different for every object
#class variable value is same for all the objects.
#self is a variable which points to the current
class Bike:
    No_tyre = 2 #class variable
    def __init__(self, name,price):
        self.bike_name= name  
        self.bike_price= price #instance variable
        print( self.bike_name ,":",self.bike_price)
obj1=Bike("pulsur",50000)
obj2=Bike("royal&field",100000)
```

```python
#Example-3: member function
# Member functions don't get called automatically; we need to call them ourselves
class Bike:
    No_tyre = 2 #class variable /Attribute/data
    def __init__(self, name,price):
        self.bike_name= name  
        self.bike_price= price #instance variable
        print( self.bike_name ,":",self.bike_price)
    def display(self):
        print("I am a Member function")
obj1=Bike("pulsur",50000)
obj2=Bike("royal&field",100000)
obj1.display()
obj2.display()
```
