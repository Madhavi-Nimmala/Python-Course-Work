# passing an arbitrary number of arguments using *args

```python
def fun(*args):
    for i in args:
        print(i)
fun()
fun("chinki")
fun("pooki","minki")
```

passing arbitrary key-value based argument using **kargs

```python
def pavan(**GF):
    for key in GF:#GF act as a dictionary
        print(key," : ",GF[key])#passing key and value
pavan(chinki = 150, miniki = 250)
pavan(chinki = 150)
pavan()
```
