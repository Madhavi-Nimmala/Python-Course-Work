#Interview questions
Q1: Is a dictionary mutable or immutable
Ans:Dictionary is mutable(insert,delete,update(values))
But keys are immutable (keys cannot be updated)
Q2.How time take to insert,delete and update(values) in dictionary?
Ans: BC, AC=O(1),WC=O(n)
Q3:Duplicate elements are allowed in dictionary or not?
ANS:Keys cannot be duplicated while values can be duplicated.
Q4:Heterogeneous items are allowed or not allowed in dictionary?
ANS:Yes Allowed
Q5:Dictionary is ordered or unordered?
ANS:It is ordered.
Before Python 3.7 version dictionary was unordered.
to make the asscess of element faster

```python
empty_dict={}
empty_dict2=dict()
print(empty_dict2)
print(type(empty_dict2))
int_dict={ 1:"mamidipandu",2:"aratipandu",3:"bathayi"}
print(int_dict)
float_dict={1.2:"mamidipandu",2.5:"aratipandu",3.7:"bathayi"}
print(float_dict)
str_dict={"A":"mamidipandu","B":"aratipandu","C":"bathayi"}
print(str_dict)
het_dict={10:"mamidipandu",20.5:"aratipandu","C":"bathayi"}#key can be int,float,str
print(het_dict)
tup_dict={(10,20):"mamidipandu",20.5:"aratipandu","C":"bathayi"}#tuple is used as key
print(tup_dict)
lis_dict={(10,20):{100,200},20.5:"aratipandu","C":"bathayi"}#list is not used key
print(lis_dict)
s1_dict={(10,20):{200,300},20.5:"aratipandu","C":"bathayi"}#set is not used key
print(s1_dict)
lis_dict={[10,20]:"mamidipandu",20.5:"aratipandu","C":"bathayi"}#list is not used key
print(lis_dict)
s1_dict={{10,20}:"mamidipandu",20.5:"aratipandu","C":"bathayi"}#set is not used key
print(s1_dict)
#tuple is used as key(immutable)
#list is not used key(mutable)
#set is not used key(mutable)
#key must be immutable
```

ACCESSING DICTIONARY ITEMS

```python
drink_items={
    "thumpsup":50,
    "sugarcane":20,
    "coconutwater":60,
    "water":20,
    "rosemilk":30
}
val=drink_items["sugarcane"]
print(val)
val2=drink_items.get("water")
print(val2)
val3=drink_items.get("rosemilk","key not available")
print(val3)
```

```python
for i in drink_items:
    print(i,drink_items[i])
```

```python
print(drink_items.keys())
```

```python
print(drink_items.values())
```

```python
print(drink_items.items())
```

```python
#nested dictionary
nested_dict={
    100:{"hello":10,"world":20},
    200:{"A":50,"B":60}
}
nested_dict[100]["world"]
```

```python
#insert & update operations on dictionary
```

```python
x_dict={"name":"chinki","age":80}
x_dict["age"]=180 #if the key is present it update
print(x_dict)
x_dict["city"]="kurnool"#if key and value not there in the dict it is insert the key and value
print(x_dict)
#insert and update using the update() function
#update() can insert and update multiple items
x_dict.update({"age":18,"occupation":"IAS","name":"minki","salary":1.5})
print(x_dict)
```

Delete the items from dictionary

```python
#pop(),popitem(),clear(),del()
```

```python
x_dict
```

```python
#pop() removes the items whose key is specified
x_dict.pop("salary")
print(x_dict)
```

```python
#popitem() will delete the last element
x_dict.popitem()
print(x_dict)
```

```python
#clear()
x_dict.clear()
print(x_dict)
```

```python
#del() 
del(x_dict)
print(x_dict)
```
