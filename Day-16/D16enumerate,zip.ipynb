#enumerate() function add the counter (index) with the list/tuple items.

```python
fruits_names=["mamidipandu","panasapandu","jamapandu"]
for index, name in enumerate(fruits_names,start=101):
    print(index,name)
```

```python
#Def: A Decorator is a function that takes another function as input,adds extra functionality and returns a modified function,
#without changing the original function code
def decorator_fun(func):
    def wrapper_fun():#wrapper function add the extra functionality
        print("buy the ingredients")
        func()
        print("add cherry")
        print("add chocolate")#extra functionality
    return wrapper_fun
def cake():
    print("take the flour")
    print("add eggs and other ingredients")
    print("put the cake in oven")
decorated_function=decorator_fun(cake)
decorated_function()
```

```python
#example-2 use of @ to represent the decorators
def decorator_fun(func):
    def wrapper_fun():#wrapper function add the extra functionality
        print("buy the ingredients")
        func()
        print("add cherry")
        print("add chocolate")#extra functionality
    return wrapper_fun
@decorator_fun
def cake():
    print("take the flour")
    print("add eggs and other ingredients")
    print("put the cake in oven")
cake()
```

```python
#example-2 use of @ to represent the decorators
def decorator_fun1(func):
    def wrapper_fun():#wrapper function add the extra functionality
        print("buy the ingredients")
        func()
    return wrapper_fun
def decorator_fun2(func):#func => cake()
    def wrapper_fun():#wrapper function add the extra functionality
        func()
        print("add cherry")
        print("add chocolate")#extra functionality     
    return wrapper_fun

@decorator_fun1
@decorator_fun2
def cake():
    print("take the flour")
    print("add eggs and other ingredients")
    print("put the cake in oven")
cake()
```
