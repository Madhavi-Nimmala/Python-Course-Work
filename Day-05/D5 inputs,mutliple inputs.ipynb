```python
#1.Concatenation using commas
#2.String concatenation using+
#3.str.format()method
#4.f-string method(python 3.6+version)p
#5.Escape sequences
#6.End and sep parametes in print()
```

```python
#1.concatenation using commas
hero_name="ramcharan"
age=41
fee=100
print(hero_name,age,fee)
```

```python
#2.string concatenation using+
hero_name="Nani"
heroine_name="nivetha thomas"
fee=100
print(hero_name +" "+heroine_name+" "+str(fee))
```

```python
#3.str.format()method
hero_name="Bob"
age=51
print("Name:{} Age:{}".format(hero_name,age))
```

```python
#3.str.format()method
hero_name="Bob"
age=51
print("Name:{1} Age:{0}".format(age,hero_name,))
```

```python
#3.str.format()method
hero_name="Bob"
age=51
print("Name:{1} Age:{0}".format(hero_name,age))
```

```python
#4.f-string method(python 3.6+version)p
hero_name="Bob"
age=51
print(f"Name:{hero_name},Age:{age}")#need variable, so keep it in the curly braces to get output
```

```python
#Escape sequences
hero_name="Bob"
age=51
fee=100
print(f"Name:\"{(hero_name)}\",\t Age:{age},\t fee:{fee}\t")#/t is used as a space and \" is used the value should be in string("")
print(f"Name:{hero_name},\n Age:{age},\n fee:{fee}\n")#\n is used excute line by line
```

```python
#6.End and sep parameters in print()
hero_name="Bob"
age=51
print(hero_name,end=" ")
print(age)
```

```python
hero_name="Bob"
age=51
fee=100
print(hero_name, age, fee, sep=" # ")
```

```python
#end parameter is used to decide how the print statement will end.
#sep parameter decides how the multiple values will be separated.
```
