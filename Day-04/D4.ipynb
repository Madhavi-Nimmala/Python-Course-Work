```python
#comparision operators(==,!=,<,>,<=,>=)
a=10
b=5
print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)
print(a<b)
print(a>b)
```

```python
#logical operators and,or,not
a=True
b=False
print(a and b)
print(a or b)
print( not a)
print(not b)
```

```python
#identity operators
a=[1,2,3]
b=[1,2,3]
c=a
print(a==b)
print(a is c)
print(a is not b)
print(a is b)
```

```python
#Membership Operators in,not in
fruits=["mamidipandu","batayi","aratipandu","panasapandu"]
print('batayi' in fruits)
print('batayi' not in fruits)
```
