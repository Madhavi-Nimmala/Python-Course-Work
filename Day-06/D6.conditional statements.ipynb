```python
#Match case is used to replace multiple if else condition
#Match case is more powerful than the switch case statement
#Match case works with variable and data structures
num=int(input("enter a number"))
match num:
    case 10:
     print("you entered 10")
    case 20:
     print("you entered 20")
    case 30:
     print("you entered 30")
    case _:
     print("something else")
```

```python
biryani=input("enter a biryani")
match biryani:
    case "paradise":
     print("Tasty and spicy")
    case "shabad":
     print("comfort and tasty")
    case "road side":
     print("uncomfort and tasty")
    case _:
     print("home biryani")
```

```python
#Example -2 passing data structure to match case
data=[1,2,3,4]
match data:
     case[4,3,2,1]:
         print("hello")
     case[1,2,3,4]:
         print("world")
      
```

```python
#example-3 if condition match with case
val=int(input("enter a number:"))
match val:
      case x if x>0:
               print("positive number")
      case y if y<0:
               print("negative number")
      case _:
               print("zero")
```
