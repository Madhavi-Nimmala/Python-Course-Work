Data Abstraction

```python
#def:Abstraction hidies internal implementation details and shows only essential features.
#Why use data abstraction:
1.To hide the complex details
2.This make the code readable and secure

#Libraries needed for data abstraction
from abc import ABC,abstract method
#Rules of data abstraction in python:
1.Abstract class can have both abstract and concrete method
2.You cannot instantiate a object from abstract class
3.All Subclass must implement ALL abstract methods.

#Abstract menthod:A method declared but not defined/implemented/without code
                 abstract method is defined as @ decorator
#Concrete method:This method have the body part
#concrete: It is not compulsory to define inside the child class

```

```python
#Example-1 How to define abstract class,abstract method
from abc import ABC, abstractmethod#abc-library file,ABC-class ,abstractmethod-decorator(@)
#abstract class
class Head_mahindra(ABC):
    @abstractmethod
    def air_bag(self):#abstractmethod
        pass
    @abstractmethod
    def Led_screen(self):
        pass
class hyd_mahindra(Head_mahindra):
    def air_bag(self):#abstractmethod
        print("I will use Nylon to design the air bag") 
        
    def Led_screen(self):
        print("sony led")
class noida_mahindra(Head_mahindra):
    def air_bag(self):
        print("I will use cotton to design the air bag")
        
    def Led_screen(self):
        print("samsung led")
obj_hyd=hyd_mahindra()
obj_noida=noida_mahindra()
obj_hyd.air_bag()
obj_hyd.Led_screen()

obj_noida.air_bag()
obj_noida.Led_screen()
```

```python
class Head_mahindra(ABC):
    @abstractmethod
    def air_bag(self):#abstractmethod
        pass
    @abstractmethod
    def Led_screen(self):
        pass
    def insurance(self):#concrete
        print("Please take mahindra company insurance")
class hyd_mahindra(Head_mahindra):
    def air_bag(self):#abstractmethod
        print("I will use Nylon to design the air bag") 
        
    def Led_screen(self):
        print("sony led")
class noida_mahindra(Head_mahindra):
    def air_bag(self):
        print("I will use cotton to design the air bag")
        
    def Led_screen(self):
        print("samsung led")
obj_hyd=hyd_mahindra()
obj_noida=noida_mahindra()
obj_hyd.insurance()
```
