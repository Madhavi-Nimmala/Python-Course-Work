Exception Handling

```python
# Def: EH allows responding to errors without crashing the running application

# With EH,we can give our own error message

```

```python
# Types of blocks in exception handling:
#1.Try-put that cide where an error can occur
#2.except-handle the error
#3.else-put code which u want to execute when there is no error
#4. finally- execute either error or not
```

## 

```python
#Example-1
try:
    print(x)
except:
    print("som error occur")
print("hello")
a,b=10,20
print (a+b)
```

```python
# example-2:
#Example-1
try:
    print(1)
except Exception as e:
    print(f"some error accour:{e}")
    print(type(e))
print("hello")
a,b=10,20
print (a+b)
```

```python
# example-2:
try:
    print(1)
except Exception as e:
    print(f"some error accour:{e}")
    print(type(e))
    print("hello")
else:
    print("hurry, we successfully excuted code")
a,b=10,20
print (a+b)
```

```python
# example-4:finally-block:execute if no error+error
try:
    y=20
    print(y)
except Exception as e:
    print(f"some error accour:{e}")
    print(type(e))
else:
    print("hurry, we successfully excuted code")
finally:
    print("i will execute whatever happen")
```

```python
# Name error: when the variable is not defined
#Example-1
try:
    print(z)

except: # this will catch all type of errors
    print("som error occur")
```

```python
# Name error: when the variable is not defined
#Example-1
try:
    print(z)
except NameError:
    print("this is name error")
except: # this will catch all type of errors
    print("som error occur")
```

```python
try:
    y=20
    print(y/0)
except NameError:
    print("this is name error")
except ZeroDivisionError: # this will catch all type of errors
    print("some ZeroDivisionError")
```

```python
try:
    print(z) # with one try we can catch one error only
    y=20
    print(y/0)
except NameError:
    print("this is name error")
except ZeroDivisionError: # this will catch all types of errors
    print("some ZeroDivisionError")
```

```python
try:
    l1=[10,20,30]
    print(l1[5])
except NameError:
    print("this is name error")
except ZeroDivisionError: # this will catch all types of errors
    print("some ZeroDivisionError")
except IndexError: # this will catch all types of errors
    print("some IndexError")
```

```python
try:
    l1=[10,20,30]
    print(l1[5])
except NameError:
    print("this is name error")
except ZeroDivisionError: # this will catch all types of errors
    print("some ZeroDivisionError")
except Exception as e:
    print(type(e),e) # to print name of the error
```
