```python
#Def:inheritance allows us to define a class that inherits the properties
#or functions of the parent class
#parent class= Base class
#Child class=  Derived class
```

```python
#Type of inheritance:
1.Single inheritance:One parent One child
2.Multiple inheritance:Many parents one child
3.Multilevel inheritance:Koduku->Nanna->Tatta->Muttata
4.Hierarichal inheritance: one parent many child
5.Hybrid inheritance:Many parent many child. Combine any two or more inheritance
```

1.Single inheitance:One parent child

```python
#example-1 structure of the single inheritance
#parent class
class hyd_food:
    taste ="very very very spicy"
    def biriyani(self):
        print("very very very tasty biriyani")
#child class
class pistahouse(hyd_food):
    pass
obj_child=pistahouse()# it is the object of child class
#child class object can access the properties and functions of parent class
print(obj_child.taste)#we can access the variable of parent class
obj_child.biriyani()
```

```python
#Example-2:Super() function
#super():used to access the parent properties and functionfrom
        #inside of the child class
#parent class
class hyd_food:
    taste ="very very very spicy"#class variable
    def biriyani(self):
        print("very very very tasty biriyani")
#child class
class pistahouse(hyd_food):
    def display(self):
        super().biriyani() # We are calling the parent class function from inside the child class function
        print(super().taste, self.taste) # We can access the parent properties and functions inside the child class function
obj_child = Pistahouse () # it is the object of child class
obj_child.display()
```

```python
#super(): Follow the  MRO approach
#MRO Method Resolution Order
class A:
    def display(self):
        print("i am A class")
        
class B:
    def display(self):
        print("I am B class")
        super().display()
class C(B,A):
    def hello(self):
        super().display()
        
obj_C=C()
obj_C.hello()
# MRO (Method Resolution Order
print([i.__name__ for i in C.__mro__])
```

```python
#The word polymorphism means "MANY FORMS"
#In polymorphism, same function or same operator can perform many different task.
#Type of Polymorphism:
#1.Function overloading
#2.operator overloading
#3.Method overriding
```

```python
1.Function Overloading
```

```python
#This type function overloading is supported in c++ and java
#Based upon number of arguments passed to function,c++ &java can differentiate the functions.
def fun(var1):
    print(var1)
def fun(var1,var2):
    print(var1, var2)
fun(10)
```

##using default arguments for function overloading

```python
def fun(var1,var2=None,var3=None):
     if var2 is not None:
         print(var1 ,var2,var3)
     else:
         print(var1)
fun(10)
fun(10,20,30)
```

```python
# example-3 : *args for function overloading
def calculation(*args):
    return sum(args)
print(calculation(100,200))
print(calculation(100,200,300))
print(calculation(100,200,300,400))
```

```python
#Example-2
def calculator(a=0,b=0,c=0,d=0):
    return a+b+c+d
print(calculator(100,200))
print(calculator(100,200,300))
print(calculator(100,200,300,400))
print(calculator(100,200,300,400))
```

```python
##function overloading using decorator
from multipledispatch import dispatch
@dispatch(int)
def fun(var1):
    print(var1)
@dispatch(int,int)
def fun(var1,var2):
    print(var1,var2)
@dispatch(float,float,int)
def fun(f1,f2,f3):
    print(f1,f2,f3)
fun(10)
fun(10,20)
fun(2.5,5.7,100)
```
